```python
# merge-sort
ar = [4, 2, 1, 3, 5, 6, 0]

def merge(left_ar, right_ar):
    if not left_ar:
        return right_ar
    if not right_ar:
        return left_ar
    output = []
    i = j = 0
    while i < len(left_ar) or j < len(right_ar):
        if j == len(right_ar) or (i < len(left_ar) and left_ar[i] < right_ar[j]):
            output.append(left_ar[i])
            i += 1
        else:
            output.append(right_ar[j])
            j += 1
    return output

def merge_sort(ar, left, right):
    if left > right:
        return []
    elif left == right:
        return [ar[left]]
    mid = left + (right - left) // 2
    left_ar = merge_sort(ar, left, mid)
    right_ar = merge_sort(ar, mid + 1, right)
    return merge(left_ar, right_ar)

print(merge_sort(ar, 0, len(ar) - 1))
```

```python
# quick sort
ar = [4, 2, 1, 3, 5, 6, 0]

def partition(ar, left, right):
    pivot = ar[right]
    i = left - 1
    for j in range(left, right):
        if ar[j] <= pivot:
            i += 1
            ar[i], ar[j] = ar[j], ar[i]

    ar[right], ar[i + 1] = ar[i + 1], ar[right]
    return i + 1


def quick_sort(ar, left, right):
    if left < right:
        pi = partition(ar, left, right)

        quick_sort(ar, left, pi - 1)
        quick_sort(ar, pi + 1, right)

quick_sort(ar, 0, len(ar) - 1)
print(ar)
```

```python
def findMaxCrossingSubArray(ar, left, mid, right):
    left_sum = -float("inf")
    acc = 0
    max_left = mid
    for i in range(mid, left - 1, -1):
        acc += ar[i]
        if acc > left_sum:
            left_sum = acc
            max_left = i

    max_right = mid + 1
    right_sum = -float("inf")
    acc = 0
    for i in range(mid + 1, right + 1):
        acc += ar[i]
        if acc > right_sum:
            right_sum = acc
            max_right = i
    return max_left, max_right, left_sum + right_sum


def findMaxSubArray(ar, left, right):
    if left == right:
        return (left, right, ar[left])

    mid = left + (right - left) // 2
    left_low, left_high, left_sum = findMaxSubArray(ar, left, mid)
    right_low, right_high, right_sum = findMaxSubArray(ar, mid + 1, right)
    cross_low, cross_high, cross_sum = findMaxCrossingSubArray(ar, left, mid, right)
    if left_sum >= right_sum and left_sum >= cross_sum:
        return (left_low, left_high, left_sum)
    elif right_sum >= left_sum and right_sum >= cross_sum:
        return (right_low, right_high, right_sum)
    return (cross_low, cross_high, cross_sum)

findMaxSubArray(ar, 0, len(ar) - 1)
```

```python
def matrix_add(A, B):
    return [[A[i][j] + B[i][j] for j in range(len(A))] for i in range(len(A))]

def matrix_sub(A, B):
    return [[A[i][j] - B[i][j] for j in range(len(A))] for i in range(len(A))]

def alphatensor_4x4_core(A, B):
    n = len(A)

    if n == 4:
        a00, a01, a02, a03 = A[0][0], A[0][1], A[0][2], A[0][3]
        a10, a11, a12, a13 = A[1][0], A[1][1], A[1][2], A[1][3]
        a20, a21, a22, a23 = A[2][0], A[2][1], A[2][2], A[2][3]
        a30, a31, a32, a33 = A[3][0], A[3][1], A[3][2], A[3][3]

        b00, b01, b02, b03 = B[0][0], B[0][1], B[0][2], B[0][3]
        b10, b11, b12, b13 = B[1][0], B[1][1], B[1][2], B[1][3]
        b20, b21, b22, b23 = B[2][0], B[2][1], B[2][2], B[2][3]
        b30, b31, b32, b33 = B[3][0], B[3][1], B[3][2], B[3][3]

        return [[sum(A[i][k] * B[k][j] for k in range(4)) for j in range(4)] for i in range(4)]

    mid = n // 2
    A11 = [row[:mid] for row in A[:mid]]
    A12 = [row[mid:] for row in A[:mid]]
    A21 = [row[:mid] for row in A[mid:]]
    A22 = [row[mid:] for row in A[mid:]]

    B11 = [row[:mid] for row in B[:mid]]
    B12 = [row[mid:] for row in B[:mid]]
    B21 = [row[:mid] for row in B[mid:]]
    B22 = [row[mid:] for row in B[mid:]]

    # C11 = A11*B11 + A12*B21
    C11 = matrix_add(alphatensor_4x4_core(A11, B11), alphatensor_4x4_core(A12, B21))
    # C12 = A11*B12 + A12*B22
    C12 = matrix_add(alphatensor_4x4_core(A11, B12), alphatensor_4x4_core(A12, B22))
    # C21 = A21*B11 + A22*B21
    C21 = matrix_add(alphatensor_4x4_core(A21, B11), alphatensor_4x4_core(A22, B21))
    # C22 = A21*B12 + A22*B22
    C22 = matrix_add(alphatensor_4x4_core(A21, B12), alphatensor_4x4_core(A22, B22))

    C = []
    for i in range(mid):
        C.append(C11[i] + C12[i])
    for i in range(mid):
        C.append(C21[i] + C22[i])

    return C

def alphatensor_large(A, B):
    if len(A) != len(B) or len(A[0]) != len(B[0]):
        raise ValueError("Matrices must be square and matching dimensions.")
    return alphatensor_4x4_core(A, B)

a = b = [
    [1,2,3,3,1,2,3,3],
    [3,2,1,1,3,2,1,1],
    [1,2,3,3,1,2,3,3],
    [1,2,3,3,1,2,3,3],
    [1,2,3,3,1,2,3,3],
    [3,2,1,1,3,2,1,1],
    [1,2,3,3,1,2,3,3],
    [1,2,3,3,1,2,3,3],
]

result = alphatensor_large(a, b)
for row in result:
    print(row)
```

```python
from collections import deque


class Color:
    RED = 0
    BLACK = 1


class Node:

    def __init__(self, data):
        self.data = data
        self.color = Color.RED
        self.left = None
        self.right = None
        self.parent = None


class RBTree:

    def __init__(self):
        self.root = None

    def get_root(self):
        return self.root

    def _rotate_left(self, x):
        y = x.right
        x.right = y.left
        if y.left is not None:
            y.left.parent = x
        y.parent = x.parent
        if x.parent is None:
            self.root = y
        elif x == x.parent.left:
            x.parent.left = y
        else:
            x.parent.right = y
        y.left = x
        x.parent = y

    def _rotate_right(self, x):
        y = x.left
        x.left = y.right
        if y.right is not None:
            y.right.parent = x
        y.parent = x.parent
        if x.parent is None:
            self.root = y
        elif x == x.parent.right:
            x.parent.right = y
        else:
            x.parent.left = y
        y.right = x
        x.parent = y

    def _fix_insert(self, k):
        while k.parent and k.parent.color == Color.RED:
            if k.parent == k.parent.parent.right:
                u = k.parent.parent.left
                if u and u.color == Color.RED:
                    u.color = Color.BLACK
                    k.parent.color = Color.BLACK
                    k.parent.parent.color = Color.RED
                    k = k.parent.parent
                else:
                    if k == k.parent.left:
                        k = k.parent
                        self._rotate_right(k)
                    k.parent.color = Color.BLACK
                    k.parent.parent.color = Color.RED
                    self._rotate_left(k.parent.parent)
            else:
                u = k.parent.parent.right
                if u and u.color == Color.RED:
                    u.color = Color.BLACK
                    k.parent.color = Color.BLACK
                    k.parent.parent.color = Color.RED
                    k = k.parent.parent
                else:
                    if k == k.parent.right:
                        k = k.parent
                        self._rotate_left(k)
                    k.parent.color = Color.BLACK
                    k.parent.parent.color = Color.RED
                    self._rotate_right(k.parent.parent)
            if k == self.root:
                break
        self.root.color = Color.BLACK

    def add(self, data):
        node = Node(data)
        y = None
        x = self.root

        while x is not None:
            y = x
            if node.data < x.data:
                x = x.left
            elif node.data > x.data:
                x = x.right
            else:
                return  # Element already exists

        node.parent = y
        if y is None:
            self.root = node
        elif node.data < y.data:
            y.left = node
        else:
            y.right = node

        self._fix_insert(node)

    def delete(self, data):
        z = self.root
        while z is not None:
            if data == z.data:
                break
            elif data < z.data:
                z = z.left
            else:
                z = z.right

        if z is None:
            return  # Node not found

        self._delete_node(z)

    def _delete_node(self, z):
        y = z
        y_original_color = y.color

        if z.left is None:
            x = z.right
            # Capture the parent before moving links
            x_parent = z.parent
            self._rb_transplant(z, z.right)
        elif z.right is None:
            x = z.left
            # Capture the parent before moving links
            x_parent = z.parent
            self._rb_transplant(z, z.left)
        else:
            # Find the successor
            y = z.right
            while y.left is not None:
                y = y.left

            y_original_color = y.color
            x = y.right

            if y.parent == z:
                x_parent = y  # y will become the parent of x
            else:
                x_parent = y.parent  # x's parent will be y's old parent
                self._rb_transplant(y, y.right)
                y.right = z.right
                y.right.parent = y

            self._rb_transplant(z, y)
            y.left = z.left
            y.left.parent = y
            y.color = z.color

        if y_original_color == Color.BLACK:
            self._fix_delete(x, x_parent)

    def _rb_transplant(self, u, v):
        if u.parent is None:
            self.root = v
        elif u == u.parent.left:
            u.parent.left = v
        else:
            u.parent.right = v
        if v is not None:
            v.parent = u.parent

    def _fix_delete(self, x, x_parent):
        # Added 'x_parent is not None' to protect against root-level crashes
        while x != self.root and x_parent is not None and (x is None or x.color == Color.BLACK):
            if x == x_parent.left:
                w = x_parent.right

                if w and w.color == Color.RED:
                    w.color = Color.BLACK
                    x_parent.color = Color.RED
                    self._rotate_left(x_parent)
                    w = x_parent.right

                if (w is None or w.color == Color.BLACK) and \
                   (w is None or w.left is None or w.left.color == Color.BLACK) and \
                   (w is None or w.right is None or w.right.color == Color.BLACK):
                    if w:
                        w.color = Color.RED
                    x = x_parent
                    x_parent = x.parent  # Safely goes up; loop condition handles if it becomes None
                else:
                    if w is None or w.right is None or w.right.color == Color.BLACK:
                        if w and w.left:
                            w.left.color = Color.BLACK
                        if w:
                            w.color = Color.RED
                        self._rotate_right(w)
                        w = x_parent.right

                    if w:
                        w.color = x_parent.color
                        if w.right:
                            w.right.color = Color.BLACK
                    x_parent.color = Color.BLACK
                    self._rotate_left(x_parent)
                    x = self.root
            else:
                w = x_parent.left

                if w and w.color == Color.RED:
                    w.color = Color.BLACK
                    x_parent.color = Color.RED
                    self._rotate_right(x_parent)
                    w = x_parent.left

                if (w is None or w.color == Color.BLACK) and \
                   (w is None or w.left is None or w.left.color == Color.BLACK) and \
                   (w is None or w.right is None or w.right.color == Color.BLACK):
                    if w:
                        w.color = Color.RED
                    x = x_parent
                    x_parent = x.parent
                else:
                    if w is None or w.left is None or w.left.color == Color.BLACK:
                        if w and w.right:
                            w.right.color = Color.BLACK
                        if w:
                            w.color = Color.RED
                        self._rotate_left(w)
                        w = x_parent.left

                    if w:
                        w.color = x_parent.color
                        if w.left:
                            w.left.color = Color.BLACK
                    x_parent.color = Color.BLACK
                    self._rotate_right(x_parent)
                    x = self.root

        if x:
            x.color = Color.BLACK

    def pprint_levels(self):
        if self.root is None:
            return

        q = deque([self.root])

        while q:
            len_level = len(q)
            level_output = []

            for _ in range(len_level):
                curr = q.popleft()
                c = "R" if curr.color == Color.RED else "B"
                level_output.append(f"{curr.data}({c})")

                if curr.left:
                    q.append(curr.left)
                if curr.right:
                    q.append(curr.right)

            print(" ".join(level_output))


tree = RBTree()
tree.add(5)
tree.add(3)
tree.add(6)
tree.add(4)
tree.add(2)
tree.delete(5)
print("Level print out [Node(Color)]:")
tree.pprint_levels()
```
